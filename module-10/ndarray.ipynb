### Numpy
SIMD -> Single Instruction Multiple Data

```python
import numpy as np

a = np.array([1, 2, 3])
a = a * 5
print(a)
```

```python
l = [1, 2, 3, 4]
arr1 = np.array(l)
print(type(arr1))

arr2 = np.array([
    [1, 2, 3, 4],
    [7, 8, 9, 10]
])

arr3 = np.array([
    [
        [1, 2, 3, 4],
        [7, 8, 9, 10]
    ], 
    [
        [11, 22, 33, 44],
        [12, 13, 14, 15]
    ]
])
```

```python
# array attributes

# dimension of the array
print(arr3.ndim) 

# shape of the array. a tuple of number of rows and column. (rows, column)
print(arr2.shape)
print(arr3.shape)

# data type of the array
print(arr3.dtype)

# size of the array
print(arr2.size)
print(arr1.size)
```

```python
"""
upcasting -> the process where NumPy automatically converts data to a higher precision or more flexible data type to prevent data loss when combining different types
""" 

a = np.array([1, 2, 2.5])
print(type(a), a)
```

#### Ndarray data type

```python
# set data type for the array

a = np.array([9, 2, 4, 5]) # int64 - default, based on your pc architecture
print(a.dtype)
a = np.array([9, 2, 4, 5], dtype=np.int8) # int8 - we're forcing numpy to set data type to int8
print(a.dtype)
```

```python
# int8 has a range of -127 to 127. what if we put value more than that and want numpy to set dtype to int8?

a = np.array([7, 4, 7, 9], dtype=np.int8) # no issues
a = np.array([128, 1, 2, 3]) # no issues, dtype = int64
a = np.array([128, 1, 2, 3], dtype=np.int8) # OverflowError 
```
