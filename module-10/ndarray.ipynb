### Numpy

```python
import numpy as np

a = np.array([1, 2, 3])
a = a * 5
print(a)
```

```python
l = [1, 2, 3, 4]
arr1 = np.array(l)
print(type(arr1))

arr2 = np.array([
    [1, 2, 3, 4],
    [7, 8, 9, 10]
])

arr3 = np.array([
    [
        [1, 2, 3, 4],
        [7, 8, 9, 10]
    ], 
    [
        [11, 22, 33, 44],
        [12, 13, 14, 15]
    ]
])
```

```python
"""
upscaling -> the process where NumPy automatically converts data to a higher precision or more flexible data type to prevent data loss when combining different types
""" 

a = np.array([1, 2, 2.5])
print(type(a), a)
```

#### Ndarray data type

```python
# set data type for the array

a = np.array([9, 2, 4, 5]) # int64 - default, based on your pc architecture
print(a.dtype)
a = np.array([9, 2, 4, 5], dtype=np.int8) # int8 - we're forcing numpy to set data type to int8
print(a.dtype)
```

```python
# int8 has a range of -127 to 127. what if we put value more than that and want numpy to set dtype to int8?

a = np.array([7, 4, 7, 9], dtype=np.int8) # no issues
a = np.array([128, 1, 2, 3]) # no issues, dtype = int64
a = np.array([128, 1, 2, 3], dtype=np.int8) # OverflowError 
```

#### Ndarray creation from existing data

```python
# list

l = [10.5, 3.1416, 20, 7.8989, 6.3333, 3.333]
a = np.array(l)
print(type(a), a.dtype, a)

a = np.array(l, dtype=np.int8) # the floating values will be gone
print(type(a), a.dtype, a)
```

```python
# mixed list

lst = [1, 3.1415, 'hello', True, False]
a = np.array(lst)
print(type(a), a.dtype, a) # dtype = <U32 referes to string data type

# tuple
t = (1, 2, 4)
b = np.array(t)
# print(b)

# set
# numpy cannot directly convert a set to an array. so, set->list->numpy.array
st = {4, 5, 4, 1, 2, 2}
c = np.array(st) # will treat the set as an object not an iterable
c = np.array(list(st)) # correct way
# print(c)
```

```python
# dictionary

dc = {'a' : 10, 'b' : 20, 'c' : 30}
keys = dc.keys()
values = dc.values()
items = dc.items()

k = np.array(list(keys))
v = np.array(list(values))
i = np.array(list(items))

print(items)
print(i.dtype)
print(i.shape)
print(i.ndim)
```

#### Ndarray generation

```python
# generate random ndarray
# syntax: array = numpy.random.randint(start, end, shape)

a = np.random.randint(1, 10, 5)
b = np.random.uniform(high=5, size=(2, 4))
print(a)
print(b)
```

```python
# generate array within a range

a = np.arange(1, 10, 1)
b = np.arange(20, 100, 5)
print(a)
print(b)
```

```python
# generate 2d array

mat = a.reshape(3, 3)
print(mat)

mat_b = b.reshape(4, 4)
print(mat_b)
```

```python
# divide a range into equal parts
# np.linespace(start, end(included), how_many_portion)

a = np.linspace(1, 10, 11) # divide the range 1 to 10 into 11 equal parts
b = np.linspace(20, 30, 2) # divide the range 20 to 30 into 2 equal parts
print(a)
print(b)
```

```python
# 1d array slicing

arr = np.arange(1, 11, 1) # generate a number between 1 to 11 with step 1
print(arr)
arr = arr[:4:2]
print(arr)
```

```python
# 2d array slicing
# syntax: arr[row_st:row_end:step, col_st:col_end:step]

arr = np.arange(1, 11, 1) # generate a number between 1 to 11 with step 1
arr = arr.reshape(2, 5) # convert the array to a 2d array with 2 rows and 5 columns
print(arr)

# getting a row
row_0 = arr[0:1, ] # output: [[1 2 3 4 5]]
row_1 = arr[1:, ] # output: [[ 6  7  8  9 10]]

# getting a column
col_0 = arr[::, 0:1,]
col_1_2 = arr[::, 1:3]
print(col_1_2)
```

```python
# slice a specific portion of a 2d matrix

a = np.arange(1, 10, ).reshape(3, 3)
b = a[1::, 1::]
print(a)
print(b)
```
