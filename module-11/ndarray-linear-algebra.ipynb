### Linear algebra

```python
import numpy as np
```

```python
a = np.arange(1, 10,).reshape(3, 3)
b = np.arange(9, 0, -1).reshape(3, 3)
print(a)
print(b)
```

```python
# dot product 
"""
while multiplying two matrices
first matrix's col no. = second matrix's row no.
a X b
(2, 3) X (3, 2)
"""
print(np.dot(a, b))
```

```python
# trace = sum of diagonal elements
print(a)

print('sum of diagonal elements:', np.trace(a))
```

```python
# determinant of a square matrix

"""
determinant is a special scalar value computed exclusively from a square matrix.
If the determinant is 0, the matrix is singular and has no inverse. 
If it is not 0, the matrix is non-singular and invertible.
"""

sq_mat = np.random.randint(1, 10, size=(3, 3))
print(np.linalg.det(sq_mat))
```
