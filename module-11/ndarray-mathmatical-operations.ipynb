### Ndarray Arithmetic Operators and Mathematical function

```python
import numpy as np
```

```python
# add/sub/mult/div values of two array

x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 20, 30, 40, 50])

add = x + y
sub = y - x
# same can be achived with functions inside numpy. .add(x, y) .subtract(y, x) ...

print(add)
print(sub)
```

```python
# comparison

x = np.array([1, 0, 3, 4, 4])
y = np.array([8, 1, 5, 4, 4])

print(np.greater(x, y))
print(np.less(x, y))
print(np.equal(x, y))
```

```python
# .all() -> are all values true?
# .any() -> is any of the value true?

print(np.greater(x, y))
print(np.less(x, y))

print(np.any(np.less(x, y)))
print(not np.all(np.greater(x, y)))
```
