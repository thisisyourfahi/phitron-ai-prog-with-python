### Searching

```python
import numpy as np
```

```python
# np.where() -> find the index(es) of a particular value

x = np.array([2, 1, 4, 9, 11, 7, 9])

index = np.where(x == 9)
print(index)

mat = np.random.randint(1, 10, size=(3, 4))
index = np.where(mat > 4)
"""
this will return a tuple of two arrays. the values in each array 
combined together represents the coordinate of that particular 
value which met the condition
"""
print(index)
```

```python
# indexes of values greater than 5

indexes = np.where(x > 5)
print(indexes)
```

```python
# make changes based on a condition
# np.where(condition, x, y) -> if condition true replace with x, otherwise y

print(x)
y = np.where(x == 9, 0, x) # if any values of x == 9, replace with 0 otherwise keep it as it is
print(y)
```

```python
# min and max value's index

print(x)
print('index of minimum value:', np.argmin(x))
print('index of maximum value:', np.argmax(x))
```
