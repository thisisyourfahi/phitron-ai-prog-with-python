### Sorting

```python
import numpy as np
```

```python
# inplace sorting -> the original array is being sorted. no copy

a = np.random.randint(1, 10, 5)
print(a)
a.sort()
print(a)
```

```python
# copy sorting
a = np.random.randint(1, 10, 5)
print('before sorting:', a)
print('sorted:', np.sort(a))
print('after sorting:', a)
```

```python
# sort in descending order

print('before sorting:', a)
print('sorted in descending order:', np.sort(a, descending=True))
```

#### 2D array sorting

```python
# sort the values row wise

a = np.random.randint(1, 20, size=(3, 4))
print('before sorting:\n', a)
print('sorted:\n', np.sort(a))
```

```python
# sort the values column wise

a = np.random.randint(10, 30, size=(3, 4))
print('before sorting:\n', a)
print('sorted:\n', np.sort(a, axis=0))
```

```python
# i want to sort the entire array

a = np.random.randint(10, 30, size=(3, 4))
print('before sorting:\n', a)
print('sorted entirely:\n', np.sort(a, axis=None).reshape(a.shape))

# explanation
# axis=None returns an 1d array. then using .reshape(a.shape), we are converting the 1d array to it's original shape.
```
