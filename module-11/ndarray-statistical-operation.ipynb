```python
import numpy as np
```

```python
# load data
# delimeter=',' -> the values are seperated by comma(,)

data = np.genfromtxt('student_scores.csv', delimiter=',', skip_header=1)
print(data)
```

```python
# data slicing

math_marks = data[::, :1]
sci_marks = data[::, 1:2]

# print(math_marks)
print(sci_marks)
```

```python
print('Minimum math marks:', np.min(math_marks))
print('Maximum math marks:', np.max(math_marks))

print('Average science marks:', np.mean(math_marks))
print('Median science marks:', np.median(math_marks))
```

```python
# standard deviation
# Standard deviation is a statistical measure that shows how much data values vary or spread out from the average

print(np.std(math_marks))
```
