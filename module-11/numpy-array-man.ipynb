### Ndarray manipulation

```python
import numpy as np
```

```python
# change dimension of an array

arr = np.random.randint(1, 100, size=(10, 5)) # 10 sample, 5 features
print(arr.shape)
arr = arr.reshape(5, 10)
print(arr.shape)
```

```python
# convert to an 1d array

row_wise_flattening = arr.flatten()
# alternative: row_wise_flattening = np.ravel(arr, order='C')
print(row_wise_flattening)

col_wise_flattening = np.ravel(arr, order='F')
print(col_wise_flattening)
```

```python
# concatenate

a = np.random.randint(1, 10, size=(2, 3))
b = np.random.randint(20, 30, size=(2, 3))

print(a)
print(b)

con_row = np.concatenate((a, b), axis=0) # axis=0 -> row wise
con_col = np.concatenate((a, b), axis=1) # axis=1 -> col wise

print(con_row)
print(con_col)
```

```python
# transpose matrix

mat = np.random.randint(1, 10, size=(2, 3))

mat_transpose = mat.T

print(mat)
print(mat_transpose)
```
