### Numpy Counting

```python
import numpy as np
```

```python
a = np.random.randint(1, 100, 50)
print(a)
```

```python
values_greater_than_27 = np.count_nonzero(a > 27)
values_greater_than_40 = np.count_nonzero(a > 40)
print('values greater than 27:', values_greater_than_27)
print('values greater than 40:', values_greater_than_40)
```

```python
# unique values

unique_values = np.unique(a) # kind of set. no duplicates. 
print(unique_values)
```

```python
# values and their count

unique_values_with_count = np.unique(a, return_counts=True)
print(unique_values_with_count)
# unique_values, unique_values_count = np.unique(a, return_counts=True)

# same can be achived with
# np.unique_counts(a)
```
