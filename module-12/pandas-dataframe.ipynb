#### Accessing data

Syntax: 
```python
DataFrame.loc[row_labels, column_labels]
```

````python
# access a particular row

print(df.loc[0])
````

````python
# access multiple rows
# syntax : loc[[rows indexes]]

df.loc[[3, 7, 0]]
````

````python
# multiple rows within a range

df.loc[11:15] # show the rows of indexes between 11 to 15
````

````python
# access a particular column

df.loc[:, 'Algorithm Marks']
````

#### Changing index and columns and iloc

````python
indexes = [x for x in range(1, 21)]
df2 = df.set_index('StudentID')
df2.head()
````

````python
# in df2 the indexes are now StudentID
# now, if you want to slice or see values of a particular range, you've use iloc(). loc() would throw 
# an error here. 

# will throw an error
# df2.loc[3:7] 

df2.iloc[3:7]
# iloc is exclusive whereas loc is inclusive. the rest is pretty same
````

````python
# rename a column
print(df.head(1))
df.rename(columns={'FullName' : 'Full Name'})
print(df.head(1))
````

````python
# no change? why? because .rename() .set_index() return a new DataFrame where changes are made. 

# solution is to set inplace=True

print(df.head(1))
df.rename(columns={'FullName' : 'Full Name', 'CompletionStatus': 'Completion Status', 'EnrollmentDate' : 'Enrollment Date'}, inplace=True)
print(df.head(1))
````

#### Drop Values

````python
# delete a particular row
df2 = df.copy()
particular_row_index = 5
df2.drop(particular_row_index, inplace=True)
df2
````

````python
# delte a particular column
df2.drop('Instructor', axis=1, inplace=True)
df2
````

````python
# change a particular property value

print(df2.loc[4])
df2.loc[4, 'Algorithm Marks'] = 87 # set the algorithm marks to 87 of index 4
print(df2.loc[4])
````

````python
# change multiple values of a 

df2.loc[4, ['Algorithm Marks', 'Completion Status']] = [89, 'Completed']
print(df2.loc[4])
````

````python
# modify values of a particular range

print(df.head())
df.loc[:4, 'Python Marks'] += 2
df.head()
````

#### Sorting in DataFrame

````python
# sort a particular column

df2.sort_values('Python Marks', inplace=True, ascending=False)
df2.head()
````

#### Filtering data based on condition

````python
# find the students whoes completion status is Not Started

not_started = df.loc[df['Completion Status'] == 'Not Started']
not_started
````

````python
# find the students who has marks above 90 in DS and completion status is completed 

completed_ds90 = df.loc[(df['Completion Status'] =='Completed') & (df['Python Marks'] >= 90)]
completed_ds90
````
