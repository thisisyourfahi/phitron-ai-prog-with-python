```python
# find matching string data

df.loc[df['City'].str.contains('New')]


# if you want to avoid case checking
# df.loc[df['City'].str.contains('new', case=False)]
```

```python
# starts with a particular string

print(df.loc[df['City'].str.contains(r'new', case=False)]) # contains new, end, start, middle. all
print(df.loc[df['City'].str.contains(r'^new', case=False)]) # starts with new only
```

```python
# ends with a particular string

print(df.loc[df['City'].str.contains(r'new$', case=False)])
```

```python
# name starts with a vowel

print(df.loc[df['Name'].str.contains(r'^[aeiou]', case=False)]) # 'Name' starts with any of the value from the list [aeiou]
```

#### Apply functions in DF

```python
mn = df['Total Marks'].min()
mx = df['Total Marks'].max()

df['Scaled Marks'] = df['Total Marks'].apply(lambda x : (x - mn) / (mx-mn))
```

```python
# custom built function

def grading_system(marks):
    if marks >= 250:
        return 'A+'
    elif marks >= 230:
        return 'A'
    else:
        return 'A'

# i want to add a column grade. where i would pass each value of totla marks and add their corespoding grade
df['Grade'] = df['Total Marks'].apply(grading_system)
df
```
