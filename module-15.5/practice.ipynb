```python
import pandas as pd
import matplotlib.pyplot as plt
```

#### 1

You have a DataFrame with column Price.
Create a new column Discounted_Price with (10% discounted price).
Save the modified DataFrame as discounted_products.csv.

```python
df = pd.read_csv('Practice.csv')
df.head()
```

```python
df['Discounted_Price'] = df['Price'] * 0.9
df.head()
```

```python
# save the df

df.to_csv('discounted_products.csv')
```

#### 2
Column: EmployeeID


Tasks:

Count duplicate entries


Remove duplicates


Display cleaned DataFrame



```python
print('Count of duplicate entries:', len(df['EmployeeID']) - len(df['EmployeeID'].unique()))

df.drop_duplicates(inplace=True)

display(df)
```

#### 3
Columns: Age, Salary

Tasks:

Count missing values


Fill missing Age with mean age


Drop rows where Salary is missing



```python
print('Count of missing age values:',df['Age'].isnull().sum())
print('Count of missing salary values:',df['Salary'].isnull().sum())

# fill missing age with mean age
df['Age'] = df['Age'].fillna(df['Age'].mean())

# drop rows where salary is missing
df.dropna(subset=['Salary'], inplace=True)

display(df)
```

#### 4
Apply Statistical Functions
Columns: Marks1, Marks2, Marks3

Tasks:

Calculate mean, max for each subject


Compute total marks per student


Find standard deviation of total marks



```python
print('Minimum of Marks1:', df['Marks1'].min())
print('Maximum of Marks1:', df['Marks1'].max())
print('Minimum of Marks2:', df['Marks2'].min())
print('Maximum of Marks2:', df['Marks2'].max())
print('Minimum of Marks3:', df['Marks3'].min())
print('Maximum of Marks3:', df['Marks3'].max())
```

```python
# calculate total marks per student

df['Total Marks'] = df.iloc[::, 5:8].sum(axis=1)
display(df)
```

```python
# standard deviation
print(df['Total Marks'].std())
```

#### 5 
Histogram, Bar, Scatter, Pie

Create:

```python
# histogram of age

plt.xlabel('Age')
plt.ylabel('Frequency of Employees')
plt.title('Employees Age Count')
plt.hist(df['Age'], edgecolor='black')
plt.show()
```

```python
# scattr plot of age vs salary

plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('Scatter plot of Employees Age with Salary')
plt.scatter(df['Age'], df['Salary'])
plt.show()
```

```python
# scattr plot of price vs discounted price

plt.xlabel('Price')
plt.ylabel('Discounted Price')
plt.title('Scatter plot of Price with Discounted Price')
plt.scatter(df['Price'], df['Discounted_Price'])
plt.show()
```

load the student_IQdata.csv
* Create a scatter plot with Study_Hours on the X-axis and IQ on the Y-axis.
* Create a scatter plot with Shoe_Size on the X-Axis and Study_Hours on the Y-axis. 
* Create a histogram of Shoe_Size

After you plot it, answer the question: Does there appear to be a correlation?


```python
df = pd.read_csv('student_IQdata.csv')
```

```python
df.head()
```

```python
plt.scatter(df['Study_Hour'], df['IQ_Score'])
plt.xlabel('Study Hours')
plt.ylabel('IQ Score')
plt.title('Study Hours VS IQ Score')

plt.show()

# positive correlation
# students with more study hours tend to have a higher iq score. although there are some outliers
```

```python
plt.scatter(df['Shoe_Size'], df['Study_Hour'])
plt.xlabel('Shoe_Size')
plt.ylabel('Study_Hour')
plt.title('Shoe Size VS Study Hour')

plt.show()

# no correlation
```

```python
plt.hist(df['Shoe_Size'])
plt.xlabel('Shoe Size')
plt.ylabel('Frequency')
plt.title('Shoe Size Frequency')

plt.show()

# the histogram is right-skewed 
# means lower shoe size have higher frequency.
```
