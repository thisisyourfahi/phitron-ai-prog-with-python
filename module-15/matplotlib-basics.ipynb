```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

#### 2D line plot
A 2D line plot is a graph that shows the relationship between two continuous variables on a flat, two-dimensional surface

```python
hours = [2, 3, 4, 1, 5, 7, 3]
days = [1, 2, 3, 4, 5, 6, 7]

plt.plot(days, hours)
```

```python
plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Study Hours in a Week')

plt.plot(days, hours)
```

```python
# plot study hours of two students

first_student_hours = [2, 3, 4, 1, 5, 7, 8]
second_student_hours = [3, 4, 5, 3, 4, 6, 7]
days = [1, 2, 3, 4, 5, 6, 7]

plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Study Hours in a Week')

plt.plot(days, first_student_hours)
plt.plot(days, second_student_hours)
```

#### 2D line plot using dataframe

```python
df = pd.read_csv('enrollment_data.csv')
```

```python
plt.xlabel('Years')
plt.ylabel('Enrollment Count')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df[['Programming', 'Digital Marketing']])
```

```python
# adding label using plt.legend()

plt.xlabel('Years')
plt.ylabel('Enrollment Count')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df[['Programming', 'Digital Marketing']], label=['Programming', 'Digital Marketing'])
plt.legend()
```

```python
# change the position of the legend()

plt.xlabel('Years')
plt.ylabel('Enrollment Count')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df[['Programming', 'Digital Marketing']], label=['Programming', 'Digital Marketing'])
plt.legend(loc='lower right')
```

```python
# .plot() takes the follwoing arguments 

# color = 'you can either pass color names or hex code'
# linewidth = numerical value
# linestyle = ['solid', 'dashed', 'dashdot', 'dotted', 'none']
# marker = ['.', 's', '^', 'v', 'D', '*', 'x', '+'] -> will mark the points of the line
```

```python
# use of marker()

plt.xlabel('Years')
plt.ylabel('Enrollment Count')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df[['Programming', 'Digital Marketing']], label=['Programming', 'Digital Marketing'], marker='*')
plt.legend(loc='lower right')
```

```python
# plot.grid()

plt.xlabel('Years')
plt.ylabel('Enrollment Count')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df[['Programming', 'Digital Marketing']], label=['Programming', 'Digital Marketing'], marker='o')
plt.legend(loc='lower right')
plt.grid()
plt.show()
```

```python
# save a graph

plt.xlabel('Years')
plt.ylabel('Enrollment Count')
plt.title('Student Enrollment over 9 Years')

plt.plot(df['Year'], df[['Programming', 'Digital Marketing']], label=['Programming', 'Digital Marketing'], marker='o')
plt.legend(loc='lower right')
plt.grid()

plt.savefig('Student Enrollment over 9 Years', dpi=300, bbox_inches='tight')
```
