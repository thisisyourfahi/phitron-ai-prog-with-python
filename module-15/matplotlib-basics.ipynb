```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

#### 2D line plot
A 2D line plot is a graph that shows the relationship between two continuous variables on a flat, two-dimensional surface

```python
hours = [2, 3, 4, 1, 5, 7, 3]
days = [1, 2, 3, 4, 5, 6, 7]

plt.plot(days, hours)
```

```python
plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Study Hours in a Week')

plt.plot(days, hours)
```

```python
# plot study hours of two students

first_student_hours = [2, 3, 4, 1, 5, 7, 8]
second_student_hours = [3, 4, 5, 3, 4, 6, 7]
days = [1, 2, 3, 4, 5, 6, 7]

plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Study Hours in a Week')

plt.plot(days, first_student_hours)
plt.plot(days, second_student_hours)
```

#### 2D line plot using dataframe

```python
df = pd.read_csv('enrollment_data.csv')
```

```python
plt.xlabel('Years')
plt.ylabel('Enrollment Count')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df[['Programming', 'Digital Marketing']])
```

```python
# adding label using plt.legend()

plt.xlabel('Years')
plt.ylabel('Enrollment Count')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df[['Programming', 'Digital Marketing']], label=['Programming', 'Digital Marketing'])
plt.legend()
```

```python
# change the position of the legend()

plt.xlabel('Years')
plt.ylabel('Enrollment Count')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df[['Programming', 'Digital Marketing']], label=['Programming', 'Digital Marketing'])
plt.legend(loc='lower right')
```

```python
# .plot() takes the follwoing arguments 

# color = 'you can either pass color names or hex code'
# linewidth = numerical value
# linestyle = ['solid', 'dashed', 'dashdot', 'dotted', 'none']
# marker = ['.', 's', '^', 'v', 'D', '*', 'x', '+'] -> will mark the points of the line
```

```python
# use of marker()

plt.xlabel('Years')
plt.ylabel('Enrollment Count')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df[['Programming', 'Digital Marketing']], label=['Programming', 'Digital Marketing'], marker='*')
plt.legend(loc='lower right')
```

```python
# plot.grid()

plt.xlabel('Years')
plt.ylabel('Enrollment Count')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df[['Programming', 'Digital Marketing']], label=['Programming', 'Digital Marketing'], marker='o')
plt.legend(loc='lower right')
plt.grid()
plt.show()
```

```python
# save a graph

plt.xlabel('Years')
plt.ylabel('Enrollment Count')
plt.title('Student Enrollment over 9 Years')

plt.plot(df['Year'], df[['Programming', 'Digital Marketing']], label=['Programming', 'Digital Marketing'], marker='o')
plt.legend(loc='lower right')
plt.grid()

plt.savefig('Student Enrollment over 9 Years', dpi=300, bbox_inches='tight')
```

#### Scatter plot

A scatter plot represents the relationship, correlation, or distribution between two numeric variables measured from the same group of items or individuals.

1. positive correlation: variable X increases, variable Y increases
2. negative correlation: variable X increases, variable Y decreases
3. no correlation: Changing X has no predictable effect on Y

**Outliers:** Isolated dots that sit far away from the rest of the data crowd, highlighting unusual anomalies or data entry errors.

```python
# load the student iq data
students = pd.read_csv('student_iq_data.csv')
```

```python
students.head()
```

```python
# let's see the correlation between study hours and iq score
plt.title('Study Hours Relation with IQ Score')
plt.ylabel('IQ Score')
plt.xlabel('Study Hours')

plt.scatter(students['Study_Hour'], students['IQ_Score'])
# positive correlation
```

```python
# let's see the correlation between show size and iq score

plt.title('Show Size Relation with IQ Score')
plt.ylabel('IQ Score')
plt.xlabel('Show Size')

plt.scatter(students['Shoe_Size'], students['IQ_Score'])
# no correlation, we shall delete the column when training our model
```

```python
# let's see the correlation between chilling hour and iq score

plt.title('Chilling Hour Relation with IQ Score')
plt.ylabel('IQ Score')
plt.xlabel('Chilling Hour')

plt.scatter(students['Chilling_Hours'], students['IQ_Score'])
# negative correlation
```

#### Histogram
histogram is a graphical representation that shows the distribution of quantitative or continuous numerical data.

```python
# frequency of iq score

plt.xlabel('IQ Score')
plt.ylabel('Frequency of Students')
plt.grid()
plt.hist(students['IQ_Score'])
```

```python
# frequency of study hours

plt.xlabel('Study Hours')
plt.ylabel('Frequency of Students')
plt.hist(students['Study_Hour'], bins=20,edgecolor='black')

# bins -> change the bin count
```

#### Bar chart

presents categorical data with rectangular bars, where the height or length of each bar is proportional to the values they represent.

```python
data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': [ 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [ 60, 88, 70, 95, 80, 98, 55],
    'Grade': [ 'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)
```

```python
df.head()
```

```python
# gender wise categorizing

gender_group = df.groupby('Gender').size()

plt.bar(gender_group.index, gender_group.values, color=['orange', 'red'], hatch=['x', 'o'])
```

```python
# load the student_data.csv

df = pd.read_csv('student_data.csv')
df.head()
```

```python
# group by completion status

completion_group = df.groupby('CompletionStatus').size()
completion_group
```

```python
# bar chart basedon comletion status

plt.bar(completion_group.index, completion_group.values, color=['green', 'orange', 'red'], edgecolor='black')
plt.xlabel('Completion Status')
plt.ylabel('Completion Count')

plt.show()
```

```python
# group by instructor
instructor_group = df.groupby('Instructor').size()
```

```python
# bar chart with instructor count

plt.bar(instructor_group.index, instructor_group.values, color=['pink', 'purple', 'violet'])
```

#### Pie Chart

represent values within 100 percent

```python
plt.pie(completion_group, labels=completion_group.index)
plt.show()
```

```python
# to show percentage, use autopct

plt.pie(completion_group, labels=completion_group.index, autopct='%1.1f%%', colors=['green', 'orange', 'red'])
plt.show()
```
