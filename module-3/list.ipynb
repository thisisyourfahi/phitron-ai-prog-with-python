Insert

```python
# adds an item in a specific position. you've to mention the index where you want to insert that item
new_list.insert(1, 10)
print(new_list)
```

Append

```python
# adds an item at the end
new_list.append(9)
print(new_list)
```

Pop and Remove

```python
# pop() -> remove the last item and returns the element
print(new_list)
# new_list.pop() 

# remove() -> removes the first matching element
new_list.remove(10)
print(new_list)
```
