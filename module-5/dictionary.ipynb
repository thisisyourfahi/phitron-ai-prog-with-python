#### Dictionary

Declaration

```python
dic = {} # empty dictionary

dic = {"first_name": "Fahiyan", "last_name": "Shah", 'age': 22, 'department': 'CSE', 'semester': 3.3, 
'cgpa': 3.92, 'contact': '01712079557', 'address': '11/2 Block D, Syedpur, Majortila, Sylhet'}

print(type(dic))
print(dic)
```

Access

```python
# dictionary['key']
print(dic['first_name'])
print(dic.get('contact'))
print(dic['address'])

# values are mutable
dic['address'] = 'Block D, Muhammudpur, Majortila, Sylhet'
print(dic['address'])
# alternative
dic.update({'contact': '01717846487'}) # update an existing key
dic.update({'university': 'Metropolitan University, Sylhet'}) # add a new key-value pair to the dictionary

# all the keys of a dictionary
print(dic.keys())
# all the values of a dictionary
print(dic.values())

# delete
dic.pop('age') # will throw an error if 'age' doesn't exist
```

```python
# iterate
for i in dic:
    print(i, ": ", dic[i])

# alternative
# for key, value in dic.items():
#     print(key, value)
```

#### Dictionary Comprehension

```python
# square values
square = {x : x ** 2 for x in range(1, 11)}
print(square)

# create a dictionary from provided list of cordinates and locations
cordinates = [(10, 10.5), (20.5, 192), (101, 102)]
locations = ['dhaka', 'chattagram', 'sylhet']

exact_location = {cor : loc for cor, loc in zip(cordinates, locations)}
print(exact_location)
```
