#### Set
* no duplicates -> duplicate items are automatically removed.
* unordered & unindexed -> my_set[0] will raise error
* immutable items -> individual items are immutable, however you can add, remove

Declaration

```python
numbers = {1, 2, 3, 4, 4, 2, 1, 5, 5} 
print(numbers)

# creating empty set
empty_set = set()
not_empty_set_but_dictonary = {}
# print(type(empty_set), type(not_empty_set_but_dictonary))

names = ['alice', 'bob', 'tommy', 'bob', 'chris, tommy']
names_set = set(names)
print(names_set) # duplicate 'bob', 'tommy' will be removed
```

Iterate

```python
for name in names:
    print(name)
```

Set modification

```python
colors = {'red', 'green'}

colors.add('blue') # adds a single element
colors.update(['yellow', 'orange', 'green']) # adds multiple elements from an iterable

colors.remove('green') # removes 'green', raises error if 'green' doesn't exists
colors.discard('purple') # removes 'purple', does not raise an error

print(colors)
```

Mathematical set operations

```python
a = {1, 2, 3, 1, 4, 2, 7}
b = {5, 9, 4, 2, 9, 4, 6}

# union
union = a.union(b) # all items from both sets
print(union)

# intersection
intersect = a.intersection(b) # only common items in both sets
print(intersect)

# difference
difference = a.difference(b) # a - b
print(difference)

# symmetric difference
symm_difference = a.symmetric_difference(b) # items that are not in both set(union - intersect)
print(symm_difference)

# disjoint
# true -> intersection = {empty}; no items in common
# false -> intersection = {items}; items in common
print(a.isdisjoint(b))
```
