```python
def area_of_triangle(height, base):
    """
    This function takes two inputs. Base and height of a triangle.
    Calculate the area and return it.
    Only works for numeric value
    """
    return (int(height) * int(base)) / 2

print(area_of_triangle(5, 10))

help(area_of_triangle) # prints the doc string.
# doc string provides a little bit of documentation for a function. 
```

Parameters vs Arguments
From a function's perspective:
* A parameter is the variable listed inside the parentheses in the function definition.
* An argument is the actual value that is sent to the function when it is called.

Arbitrary Arguments - *args
* If you do not know how many arguments will be passed into your function, add a * before the parameter name.

```python
def adder(*numbers):
    sum = 0
    for n in numbers:
        sum += n
    return sum

print(adder(1, 2, 3))
print(adder(10, 20, 50, 50, 10, 5, 5))
```

##### Arbitrary Keyword Arguments - **kwargs
* The **kwargs parameter allows a function to accept any number of keyword arguments.
* Inside the function, kwargs becomes a dictionary containing all the keyword argume

```python
def my_info(**kwargs):
    """
    You can pass any number of keyword arguments.
    keyword argument means: 
    syntax: keyword = value
    example: name = 'fahiyan'
    """
    for key,value in kwargs.items():
        print(f'{key}: {value}')

my_info(name = 'Fahiyan', age = 22, gender = 'Male')
my_info(name = 'Arpita Bhadra', age = 23, department = 'LLB', village = 'Shahporan') 
# it does not matter how many keyword arguments you sent in your last call. 
```

Lambda
