The `open()` function is used for file handling. 

It usually takes two parameters; filename and mode

common modes are: 
1. 'r' - read
2. 'a' - append
3. 'w' - write
4. 'x' - create

```python
# open the sample.txt file, throws an error if the file doesn't exist
file = open('sample.txt', 'r')
content = file.read()
print(content)
```

```python
# readline() -> used to read a single line
file = open('sample.txt', 'r')
x = file.readline()
print(x)

# readlines() -> returns a list of sentences
lines = file.readlines()
print(type(lines))
print(lines)
```

#### Close Files

```python
# it's a good practice to always close the file when you're done with it.

# manually closing
print(file.closed) # False -> because the file hasn't been closed
file.close()
print(file.closed)
```

```python
# alternative approach
# with statement takes care of closing the file automatically

with open('sample.txt', 'r') as f:
    print(f.read())
    print(f.closed) # False -> because you're still inside with

print(f.closed) # True -> outside with, the file doesn't exist
```
