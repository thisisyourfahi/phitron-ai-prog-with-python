The `open()` function is used for file handling. 

It usually takes two parameters; filename and mode

common modes are: 
1. 'r' - read
2. 'a' - append
3. 'w' - write
4. 'x' - create

#### Read

```python
# open the sample.txt file, throws an error if the file doesn't exist
file = open('sample.txt', 'r')
# file.read(10) # it will read the string upto 10th index
content = file.read()
print(content)
```

```python
# readline() -> used to read a single line
file = open('sample.txt', 'r')
x = file.readline()
print(x)

# readlines() -> returns a list of sentences
lines = file.readlines()
print(type(lines))
print(lines)
```

#### Close Files

```python
# it's a good practice to always close the file when you're done with it.

# manually closing
print(file.closed) # False -> because the file hasn't been closed
file.close()
print(file.closed)
```

```python
# alternative approach
# with statement takes care of closing the file automatically

with open('sample.txt', 'r') as f:
    print(f.read())
    print(f.closed) # False -> because you're still inside with

print(f.closed) # True -> outside with, the file doesn't exist
```

#### Write

```python
# creating a new file and writing here
with open('sample-write.txt', 'w') as file:
    file.write('hello sample write\n')
    file.write('I am learning python file-handling.')
```

```python
# writing on an existing file, it will overwrite the previous writing
with open('sample-write.txt', 'w') as file:
    file.write('My name is Fahiyan\n')
    file.write('Learning is going on very fast.\n')
    file.write('Learning python is pretty fun.\n')
```

```python
# no overwriting, just append my new text below the previous text
with open('sample-write.txt', 'a') as file:
    file.write('This text has been added to the file with append mode. The previous text is untouched.\n')
```

```python
# reading the file that we created above
with open('sample-write.txt', 'r') as file:
    contents = file.readlines() # this will return a list of lines from the text file.
    contents = list(map(str.strip, contents)) # removing the '\n' chararcter from each line.
    for line in contents:
        print(line)
```

```python
# writing the contents of a list into a file
sentences = [
    "The quick brown fox jumps over the lazy dog.",
    "Artificial intelligence is rapidly transforming modern industries.",
    "She walked through the silent forest as the sun began to set.",
    "Learning a new language opens up countless opportunities.",
    "Technology continues to bridge distances between people across the globe.",
    "A balanced diet and regular exercise are essential for good health.",
    "The old bookstore on the corner held secrets of a bygone era.",
    "Music has the unique power to evoke deep emotions.",
    "Innovation requires both creative thinking and disciplined execution.",
    "Every challenge presents a hidden opportunity for growth.",
]

with open('sentences-from-list.txt', 'w') as file:
    for sentence in sentences:
        file.write(f'{sentence}\n')
```

#### Cursor/Pointer

```python
# .tell() -> tells where the cursor is at now currently
# .seek(x) -> move the cursor to xth index
with open('sentences-from-list.txt', 'r') as file:
    print(f'Cursor index: {file.tell()}') 
    x = file.readline()
    print(f'{x} Length of the string: {len(x)}, Cursor index: {file.tell()}')
    x = file.readline()
    print(f'{x} Length of the string: {len(x)}, Cursor index: {file.tell()}')
    file.seek()
```
