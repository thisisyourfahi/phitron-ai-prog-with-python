#### Problem 1:
Get the text from sample.txt , and count the following

Number of lines

Number of words

Number of characters

and save it to the counter_of_string.txt file

```python
from functools import reduce

with open('sample.txt', 'r') as file:
    content = file.readlines()
    lines = list(map(str.strip, content))
    num_of_lines = len(lines)

    words_in_each_line = list(map(lambda x : len(x.split()), lines)) # [2, 2, 2, 4]
    number_of_words = sum(words_in_each_line)
    # alternative
    # number_of_words = reduce(lambda x, y : x + y, words_in_each_line)
    
    words = []
    for line in lines:
        x = line.split()
        for i in x:
            words.append(i)
    
    num_of_char = [len(x) for x in words]
    num_of_char = sum(num_of_char)
    # print(num_of_char)

    with open('count-of-sample.txt', 'w') as f:
        f.write(f'Number of lines: {num_of_lines}\n')
        f.write(f'Number of words: {number_of_words}\n')
        f.write(f'Number of characters: {num_of_char}\n')
```
