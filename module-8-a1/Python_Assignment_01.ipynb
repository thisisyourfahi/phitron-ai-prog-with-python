## Question 1 — Divisible by 4 but not 8 (20 Marks)

Given a list of integers, extract all numbers that are divisible by 4 but not divisible by 8.

**Input:** Space-separated integers  
**Output:** A list of numbers satisfying the condition

**Example**
- Input: `4 6 8 12 16 20 24 28`
- Output: `[4, 12, 20, 28]`

## Question 2 — Compute Ratio With Exception Handling (20 Marks)

Write a function `compute_ratio(x, y)` that returns `(x + y) / (x * y)`. Use exception handling to manage division by zero.

**Input:** Two values  
**Output:** The computed result or an error message

**Example**
- Input: `5 0`
- Output: `Error: Division not possible`

## Question 3 — Find the Longest Word in a File (20 Marks)

Write a Python program that reads a text file and finds the longest word.

**Requirements:**
- Open and read data from a file (for example, `data.txt`).
- Ignore punctuation marks.
- If multiple words share the maximum length, printing any one is acceptable.

**Input:** A text file containing multiple lines of text  
**Output:** The longest word

**Example file content (`data.txt`):**
```text
Python is a wonderful language for machine learning.
```
**Expected output:** `wonderful`

## Question 4 — Words That Start With a Vowel (20 Marks)

Convert a sentence into a list containing only words that start with a vowel (`a`, `e`, `i`, `o`, `u`), ignoring case.

**Input:** A single-line sentence  
**Output:** A list of strings

**Example**
- Input: `An apple a day keeps the doctor away`
- Output: `['An', 'apple', 'a', 'away']`

## Question 5 — Palindrome Checker (20 Marks)

A word is a palindrome if reversing its letters results in the same word, ignoring case. Write a program to check whether a given string is a palindrome.

**Input:** A single word string  
**Output:** `Palindrome` or `Not Palindrome`

**Example**
- Input: `Radar`
- Output: `Palindrome`
