```python
def compute_ratio(x, y):
    try:
        ratio = (x + y) / (x * y)
    except ZeroDivisionError as e:
        print('Error: Division not possible')
    else:
        return ratio

compute_ratio(10, 0)
```

```python
def removePunctuation(word):
    w = word[-1]
    if (ord(w) >=65 and ord(w) <= 90) or (ord(w) >= 97 and ord(w) <= 122):
        return word
    else:
        return word[:len(word) - 1]

def longestWord(content):
    words = content.split()
    words = [removePunctuation(w) for w in words]

    word = words[0]
    max_len = len(word)
    for w in words:
        if len(w) > max_len:
            max_len = len(w)
            word = w
    return word

with open('demo.txt', 'r') as file:
    content = file.read()
    print(longestWord(content))
```

```python
def isVowel(c):
    return c == 'a' or c == 'e' or c == 'i' or c == 'o' or c == 'u'

def solution(sentence):
    words = sentence.split()
    words_with_vowel = [w for w in words if isVowel(w[0].lower())]
    return words_with_vowel

print(solution('An apple a day keeps the doctor away'))
```

```python
def isPalindrome(word):
    word = word.lower()
    rev_word = word[::-1]
    if word == rev_word:
        return 'Palindrome'
    else:
        return 'Not Palindrome'
    
print(isPalindrome('cat'))
print(isPalindrome('Radar'))
print(isPalindrome('mam'))
```
