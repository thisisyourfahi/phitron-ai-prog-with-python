### Class and Objects

```python
class Phone:
    category = 'Electronics' # class attribute

    def __init__(self, model, battery, camera, battery_percentage = 100):
        self.model = model # instance attribute
        self.battery = battery
        self.camera = camera
        self.battery_percentage = battery_percentage
    
    # methods
    def charge(self):
        self.battery_percentage = 100
        print('Battery has been fully charged!')
    
    def capture(self, total_photo):
        if(self.battery_percentage <= 0) :
            print('Phone Dead!!!')
        else:
            self.battery_percentage -= total_photo
            print(f'{total_photo} Photo captured.')
```

```python
# creating objects
apple = Phone('iPhone 18', 4200, 12)
redmi = Phone('Note 13 Pro Plus', 5000, 48)

print(apple.battery_percentage)
apple.capture(10)
print(apple.battery_percentage)
apple.capture(20)
print(apple.battery_percentage)
```

### Single Inheritance

```python
class SmartPhone(Phone):
    def __init__(self, model, battery, camera, processor):
        super().__init__(model, battery, camera)
        self.processor = processor

    def charge(self):
        print('Fast charging enabled!')
        super().charge()
```

```python
pro = SmartPhone('Galaxy S26', 5000, 100, 'Snap Dragon')
pro.capture(30)
print(pro.battery_percentage)
pro.charge()
print(pro.battery_percentage)
```

### Multiple Inheritance

```python
class GamingPhone:
    def __init__(self, cooling, rgb):
        self.cooling = cooling
        self.rgb = rgb
```

```python
class AllRounderPhone(SmartPhone, GamingPhone):
    def __init__(self, model, battery, camera, processor, cooling, rgb):
        SmartPhone.__init__(self, model, battery, camera, processor)
        GamingPhone.__init__(self, cooling, rgb)
```

```python
# black = AllRounderPhone('Red Magic 8', 8000, 40, 'Snap Dragon 8 Elite', 'Liquid Cooling', True)

black = AllRounderPhone('RM 8 Pro', 8000, 40, 'SD 8 Elite', 'Liquid', True)

print(black.battery_percentage)
black.charge()
print(black.cooling)
print(black.rgb)
```
