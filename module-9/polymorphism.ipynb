### Polymorphism

```python
class Camera:
    def __init__(self, name):
        self.name = name
    
    def capture(self):
        print('A photo is captured')
```

```python
class Smart_Camera(Camera):
    def __init__(self, name, res):
        super().__init__(name)
        self.res = res

class Drone(Camera):
    def __init__(self, name, camera_count):
        super().__init__(name)
        self.camera_count = camera_count

class DSLR(Camera):
    def __init__(self, name, lense, tripod):
        super().__init__(name)
        self.lense = lense
        self.tripod = tripod
```

```python
phone = Smart_Camera('Xiomi', 200)
dslr = DSLR('Canon 6D', True, True)
drone = Drone('DJI 4 Pro', 3)

# all of these instances have the same output. but for different instance you want different behavior. Polymorphism
phone.capture()
dslr.capture()
drone.capture()
```

```python
class Smart_Camera(Camera):
    def __init__(self, name, res):
        super().__init__(name)
        self.res = res

    def capture(self):
        print(f'A picture has been captured with your {self.name} device.')

class Drone(Camera):
    def __init__(self, name, camera_count):
        super().__init__(name)
        self.camera_count = camera_count

    def capture(self):
        print(f'A high res video has been recorded with your {self.name} drone.')

class DSLR(Camera):
    def __init__(self, name, lense, tripod):
        super().__init__(name)
        self.lense = lense
        self.tripod = tripod

    def capture(self):
        print(f'A high res portrait has been taken with your {self.name} device.')
```

```python
phone = Smart_Camera('Xiomi', 200)
dslr = DSLR('Canon 6D', True, True)
drone = Drone('DJI 4 Pro', 3)

# this time, different output. Polymorphism - different form. same name, but different form
phone.capture()
dslr.capture()
drone.capture()
```
